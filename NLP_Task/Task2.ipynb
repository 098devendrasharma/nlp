```python
import spacy
import re
import pandas as pd
# Load English NLP model
nlp = spacy.load("en_core_web_sm")
```

```python
doc_text = """
This is Rahul Verma.
    Previously employed at TCS and Accenture.
    Total experience is 5 years.
    Skilled in SQL, Power BI, and Data Analysis.
    Location: Mumbai.
    Email: rahul.verma@outlook.com 
    """
```

```python
doc_text = [
    """
    This is Rahul Verma.
    Previously employed at TCS and Accenture.
    Location: Mumbai.
    Email: rahul.verma@outlook.com 
    """,
    """
    Jane Doe is a data scientist.
    She used to work at Google and Meta.
    Current Location: San Francisco.
    Contact her at jane.doe@gmail.com
    """
]

all_records = []

for text_profile in doc_text:
    doc = nlp(text_profile)
    
    current_entry = {
        "Name": None,
        "Companies": [],
        "Location": None,
        "Email": None
    }
    
    for ent in doc.ents:
        if ent.label_ == "PERSON" and not current_entry["Name"]:
            current_entry["Name"] = ent.text
        elif ent.label_ == "ORG":
            current_entry["Companies"].append(ent.text)
        elif ent.label_ == "GPE" and not current_entry["Location"]:
            current_entry["Location"] = ent.text

    for token in doc:
        if token.like_email:
            current_entry["Email"] = token.text
            break
            
    if current_entry["Companies"]:
        current_entry["Companies"] = ", ".join(current_entry["Companies"])
    else:
        current_entry["Companies"] = None

    all_records.append(current_entry)

df = pd.DataFrame(all_records)

print(df)
```
