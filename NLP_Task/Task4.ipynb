# FastText

FastText is an open-source library developed by Meta Platforms for efficient text classification and word representation (word embeddings). It is widely used in Natural Language Processing (NLP) tasks.

## What FastText Does

FastText helps computers understand text by converting words or sentences into numerical vectors called **embeddings**. These embeddings capture semantic meaning, so similar words have similar vector representations.

### Examples
- “king” and “queen” will have nearby embeddings.
- “apple” and “banana” will also be close because both are fruits.

## Key Features

- Fast training and efficient performance
- Supports multiple languages
- Handles rare and misspelled words using subword information

## Applications

FastText is useful for:
- Text classification
- Sentiment analysis
- Word similarity
- Language identification
- Search and recommendation systems

```python
# Train FastText model and create embeddings
import fasttext

model = fasttext.train_unsupervised(
    r"..\gameofthrones\third.txt",
    model='skipgram',
    dim=100,
    ws=5,
    epoch=10,
    lr=0.05,
    minCount=2
)
# Get word embedding
embedding = model.get_word_vector("Blane")

print(embedding)
```

```python
print(len(embedding))
```

```python
model.get_nearest_neighbors("Blane")
```

```python
import pandas as pd 
df = pd.read_csv(r"..\content\IMDB Dataset - IMDB Dataset.csv")
df.head()
```

```python
# I will take 1 rows 1 column and store it on text variable
text = df.iloc[0, 0]
text = text.lower()
print(text)
```

```python
# apply the remove HTML Tags
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'', text)
text = remove_html_tags(text)
print(text)
```

```python
# Remove punctuations in the text
import string
text = text.translate(
    str.maketrans("", "", string.punctuation)
)

print(text)
```

```python
# Extract NER Name Entity Recognition 
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text) 
```

```python
for ent in doc.ents:
    print(ent.text, " -- > ", ent.label_)
```

```python
from spacy import displacy
displacy.render(doc,style= "ent",jupyter=True)
```
