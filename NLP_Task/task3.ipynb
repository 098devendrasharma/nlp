```python
import pandas as pd 
```

```python
df = pd.read_csv(r"..\content\IMDB Dataset - IMDB Dataset.csv")
```

```python
df.head()
```

```python
text = df.iloc[0, 0]
text = text.lower()
text
```

### Remove HTML Tag

```python
# Remove HTML tag 
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'', text)
text = remove_html_tags(text)
```

### Remove Punctuation


```python
import string
text = text.translate(
    str.maketrans("", "", string.punctuation)
)

print(text)
```

### NER Extract

```python
import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
from spacy import displacy
displacy.render(doc,style= "ent",jupyter=True)
```

## POS Tagging

```python
for token in doc:
    print(
    "Word:", token. text,
    "| POS:", token.pos_,
    "| TAG:", token.tag_
    )
```
