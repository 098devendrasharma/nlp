# NER --- >
```
Named Entity Recognition (NER) is an NLP technique used to find important names and information in text-such as people, places, organizntions, dates, and amounts-and Label them with their correct # type.

In short: NER helps a machine understand "who", "what", and "where" from text.

ExampLe (Very Simple)
    Text: "Rohit Sharma Lives in Mumbai."

NER Output:
    Rohit Sharma - PERSON
    Mumbai - LOCATION (GPE)

Why Do We Use NER?
    We use NER to convert unstructured text into meaningful, structured data.
    Main Reason : 
        1. Automatically extract important information.
        2. Reduce mannual work in reading document.
        3. Help machine understand context, not just keyword 
        4. Make text usable for analytics , ML models and database.
```

## Load NER MODEL

````python
import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")
````

````python
# eX-1
text = "Barack Obama was born in Hawaii and became President of the United States."
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
````

````python
# ex - 2
text = "Google CEO Sunder Pichai announced a $10 Billion investment in India on 15th July 2020"

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
````

````python
# ex - 3 Person and location
text = "Sachin Tendulkar was born in Mumbai"

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
````

````python
# ex - 4 Organization and Date
text = "Microsoft was founded in 1975 by Bill Gates"

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
````
