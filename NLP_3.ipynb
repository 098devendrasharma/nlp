## Text Summarization

### 1. Bag of Words

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'text' : ['people watch cricket', 'player was player',
                              'people write comment', 'cricket play comment'],
                   'output' : [1,1,0,0]})
```

```python
df
```

```python
## Unique words --> people watch cricket player was write comment play
# [11100000]
# [00021000]
# [10000110]
# [00100011]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()

BOW = cv.fit_transform(df['text'])

# Vocab
print(cv.vocabulary_)
```

```python
print(BOW[0].toarray())
print(BOW[1].toarray())
```

```python
cv.transform(["cricket watch and write comment of cricket comment"]).toarray()
```

### 2. N-Grams

```python
# vocab
print(v.vocabulary_)
```

### 3. TF-IDF (Term Frequency - Inverse Document Frequency)

```python
# TF(t, d) = (number of occurance of term t in document d) / (total number of terms in documnt d)

# Example --> d1 = people watch cricket
# tf(people, d1) = 1 / 3

# d2 = match was match
# tf(match, d2) = 2 / 3

# IDF(t) = log(Total number of documents in corpus / Number of documents with term t in them)
# Here corpus means number of sentences

# Example ==>
# d1 = people watch cricket.
# d2 = cricket was cricket.
# d3 = people write comment.
# d4 = cricket write comment.

# idf(cricket) = log(4/4)
# idf(people) = log(4/2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()

tfidf.fit_transform(df['text'])
```

```python
tfidf.fit_transform(df['text']).toarray()
```

```python
print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```

## Dataset -> IMDB Dataset

```python
df = pd.read_csv('/content/IMDB Dataset - IMDB Dataset.csv')
df.head()
```

```python
## Lowercasing
data = df.loc[0:20].copy()
data['review'] = data['review'].str.lower()
```

```python
## Removing HTML tags
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'', text)

data['review'] = data['review'].apply(remove_html_tags)
```

```python
## Removing URL's
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www\.\S+')
  return pattern.sub(r'', text)
data['review'] = data['review'].apply(remove_url)
```
